```python
from sklearn.datasets import fetch_california_housing
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = fetch_california_housing(as_frame = True)
df = data.frame
```

```python
print(df.head())
print("Shape: ", df.shape)
print(df.describe())
print(df.isnull().sum())
```

```python
plt.scatter(df["MedInc"], df["MedHouseVal"], alpha = 0.1, s=5)
plt.xlabel("Median income (×$10k)")
plt.ylabel("Median house value (×$100k)")
plt.title("Income vs house price")
plt.show()
```

```python
X = df.drop(columns=["MedHouseVal"]).values
y = df["MedHouseVal"].values

rng = np.random.default_rng(42)
idx = rng.permutation(len(X))
split = int(0.8 * len(X))
train_idx, test_idx = idx[:split], idx[split:]

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]
print(X_train.shape, X_test.shape)
```

```python
mean = X_train.mean(axis = 0)
std = X_train.std(axis = 0)

X_train_s = (X_train - mean)/std 
X_test_s = (X_test - mean)/std

print(X_train_s.mean(axis = 0).round(2))
print(X_train_s.std(axis = 0).round(2))
```

```python
n, d = X_train_s.shape
weight = np.zeros(d)
bias  = 0.0
learnRate = 0.1
epochs = 1000
losses = []

for epoch in range(epochs):
   y_pred = X_train_s @ weight + bias 

   error = y_pred - y_train
   loss = np.mean(error ** 2)
   losses.append(loss)

   dw = (2/n) * (X_train_s.T @ error)
   db = 2 * np.mean(error)

   weight -= learnRate * dw
   bias -= learnRate * db

   if epoch % 100 == 0:
      print(f"epoch {epoch:4d} loss {loss:.4f}")

print("weights: ", weight.round(3))
print("bias: ", round(bias, 3))
```

```python
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Training MSE")
plt.title("Loss going down = model learning")
plt.show()
```

```python
y_test_pred = X_test_s @ weight + bias

mse = np.mean((y_test_pred - y_test) ** 2)
rmse = np.sqrt(mse)
r2 = 1 - np.sum((y_test - y_test_pred) ** 2) / np.sum((y_test - y_test.mean()) ** 2)

print(f"Test MSE:  {mse:.4f}")
print(f"Test RMSE: {rmse:.4f}  (≈ ${rmse*100:.0f}k average error)")
print(f"Test R²:   {r2:.4f}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

model = LinearRegression()
model.fit(X_train_s, y_train)
sk_pred = model.predict(X_test_s)

print("sklearn RMSE:", round(np.sqrt(mean_squared_error(y_test, sk_pred)), 4))
print("sklearn R²:  ", round(r2_score(y_test, sk_pred), 4))
print("sklearn weights:", model.coef_.round(3))
print("your weights:   ", weight.round(3))
```
